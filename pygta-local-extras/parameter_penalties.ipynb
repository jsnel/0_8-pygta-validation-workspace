add parameter_penalties to the pyglotaran model. an example of a definition (to be added to a model.yml file) is in the cell below, where equal.1 is a parameter equal to 1. note that this is much simpler than clp_penalties. in the result of the optimization a row analogous to the "RMSE additional penalty" should be added.


parameter_penalties:

- type: equal
  source: ratenolhcb16a.kclose65
  target: ratecol0a.kclose65
  parameter: equal.1
  weight: 1e3


analogous to the toy example in the cell below (where the names of the parameters are p1 and p2), penalty residuals should be added.


````python
import numpy as np
from scipy.optimize import least_squares


def objective_function(params, x_data, y_data, weight):
    # Unpack parameters (assuming p1 is at index 0 and p2 is at index 1)
    p1 = params[0]
    p2 = params[1]

    # 1. Calculate standard data residuals (e.g., a simple linear model: y = p1*x + p2)
    model_predictions = p1 * x_data + p2
    data_residuals = model_predictions - y_data

    # 2. Calculate the penalty residuals (taking the square root of the weight)
    sqrt_w = np.sqrt(weight)
    penalty_1 = sqrt_w * (p1 / p2 - 1)
    penalty_2 = sqrt_w * (p2 / p1 - 1)

    # 3. Append penalties to the residuals array
    # The solver will square all elements in this array and sum them up
    return np.append(data_residuals, [penalty_1, penalty_2])


# --- Example Usage ---
# Dummy data
x_data = np.array([1.0, 2.0, 3.0])
y_data = np.array([2.0, 4.0, 6.0])

# Initial guess for [p1, p2, ...]
initial_guess = [1.5, 1.5]
penalty_weight = 10.0

# Run optimization
result = least_squares(objective_function, x0=initial_guess, args=(x_data, y_data, penalty_weight))

print("Optimized parameters:", result.x)
````
